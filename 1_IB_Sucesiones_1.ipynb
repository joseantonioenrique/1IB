## Introducción

Los conjuntos, las series y las secuencias están interrelacionados. Cada uno de ellos es útil para tratar grupos de números, especialmente grandes grupos de números. Los conjuntos pueden contener datos útiles y ayudar a los matemáticos a comprender estos grupos de números con mayor claridad. Las secuencias siempre forman patrones. Estos patrones tienen muchas aplicaciones prácticas en otras áreas de las matemáticas y también en la vida. De estas secuencias surgen proporciones y fórmulas que tienen muchas aplicaciones. Una de ellas es la secuencia de Fibonacci, que está directamente relacionada con la proporción áurea. Esta proporción se utilizaba a menudo en la arquitectura antigua y se sigue utilizando hoy en día.


Una **sucesión** es una lista infinita de números definida por una regla. Los números de la sucesión suelen escribirse como $a_1$, $a_2$, $a_3$,\dots Los puntos significan que la lista continúa para siempre. Un ejemplo sencillo es la sucesión
$$a_{1},a_{2},a_{3},\cdots,\; a_{n},\;a_{n+1},\cdots $$




*  Los números en la sucesión, se dice que son sus miembros o términos.
*  Una sucesión puede ser **finita** o **infinita **

sucesion_1.png

Debemos tener claro dos conceptos bien diferenciados. El **lugar que ocupa el término**, normalmente indicado por $n$, y el valor del término, indicado por $a_{n}$.
En este caso hablamos de $n$ como el número de filas que tiene el triángulo, y $a_{n}$ el número de bolas que hay en él.
$$
\begin{array}{cc}
Filas & Número\; de\; bolas\\
1     & 1\\
2     & 3\\
3     & 6\\
\cdots & \cdots \\
n  & ?\\
\end{array}
$$
En la sucesión anterior, el lugar $4$, es decir $n=4$, 4 filas, es el cuarto triángulo, y el número de bolas que tiene el cuarto triángulo, es el valor del término 4, en este caso, $10$.

¿Podrías saber cuántas bolas tienes con 20 filas?. ¿Y 200?. Necesitamos lo que llamamos, **un término general**![alturas.gif](![alturas.eps](foto.jpg))

Veamos otro ejemplo,

* En la lista $4,7,10,13, \dots$ el término $5$ sería el $16$, pues habrás observado que va sumando $3$ cada vez- Una sencilla cuenta nos ayuda a conocer el valor del término, ahora bien, ¿qué operaciones debes realizar para conocer el término que ocupa el lugar 100 en la sucesión, es decir, $a_{100}$?. Observa que para llegar al que ocupa el lugar $4$, por ejemplo, hemos sumado a partir del primero, el $4$, tres veces el $3$, para llegar al lugar $5$ hemos sumado $4$ veces el 3
$$16=4+3+3+3+3=4+{\red 4}\cdot 3$$
para obtener el de lugar $6$, es decir $a_6$ sumamos, una menos, 5 veces el valor fijo que se suma, **la diferencia**, el $3$
$$19=4+3+3+3+3+3=4+{\red 5}\cdot 3$$
En general, para obtener el lugar $n$ tendremos que sumar $n-1$ vez la diferencia, es decir, $4+(n-1)\cdot 3$.

Ahora vamos a crear un programa que nos ayude a calcular esto. El lenguaje que usaremos para comunicarnos con la maquina y pedirle los cálculos se llama Python. Hay muchos lenguajes diferentes, C++, java, javascript, pascal,c#,... pero todos guardan una misma idea, comunicarnos con la maquina.  
Para ejercutarla solo debes dar al boton de play.

```python
#esto son comentarios que la maquina no ejecuta, sirve para dejarme mensajes y saber lo que estoy haciendo.
# Tomanos una expresión general de la sucesion sucesión 3(n-1)+4, en este caso será una función, cuya entrada es el valor n, posición, y la salida será el
# valor que ocupa la posición n
# para ello definimos la fórmula que llamamos sucesion (sin tilde, in english) y esta fórmula depende del valor n
# le daremos un valor a n y luego lo calculamos.
sucesion = lambda n: 3*(n-1)+4

# Vamos a pedirle al usuario que nos diga cuál es el valor de la posición que quiere conocer.
# El usuario introduce un número y este se almacena, como un entero int(), en la variable num, de forma que num valdrá lo que el usario introduzca.
num = int(input("Introduzca cuál es el valor de la posición cuyo valor desea conococer: "))
# Vamos a calcular el valor del número que corresponde a la sucesión en el lugar que indica el usario. Lo calculamos y después lo mostramos.
print("La posición " +str(num)+ "está ocupado por el número cuyo valor es: ", sucesion(num))

```

## Listamos los términos, iteraciones en Python
Vamos a solicitar al programa que nos muestre los $10$ primeros términos de una sucesión. Para ellos vamos a usar un ejemplo práctico en física.
Si hacemos rebotar una pelota contra un tablero horizontal, en caida libre, observamos que la velocidad perpendicular al tablero va disminuyendo mientras que la paralela queda inalterada.
$$\begin{array}{l} v_x=u_x\\ v_y=-e\cdot u_y\end{array}$$

donde $e$ es el coeficiente de restitución, un parámetro que mide la conservación de la energía tras un choque. (en la mayoría de los casos este coeficiente es menor que 1).

Vamos a suponer que la pelota se lanza desde una altura $h$.

alturas.png

Entonces por el principio de conservación de la energía tenemos
$$mgh=\dfrac{1}{2}\cdot m \cdot u_{1}^2\qquad despejando \; u_{1}^2=\dfrac{2mgh}{m}\; \rightarrow\; u_{1}=\sqrt{2gh}$$
cuando la pelota choque con el suelo se produce un movimiento ascendente con una velocidad inicial $v_1=e\cdot u_{1}$, nuevamentes por el principio de conservación de la energía, despejando y teniendo en cuenta el valor anterior, llegamos a que
$$mgh_1=\dfrac{1}{2}mgv_{1}^2\qquad h_1=e^2\cdot h$$

Si repetimos el proceso, en el segundo rebote nuevamente por el prinicipio de conservación de la energia y despejando

$$mgh_2=\dfrac{1}{2}mgv_{2}^2\qquad h_2=e^2\cdot h_1=e^4\cdot h$$

y por tanto llegamos a que
$$h_n=e^{2n}\cdot h$$

Vamos a calcular las alturas alcanzadas por una pelota lanzada desde $2m$  de altura con un coeficiente de restitución de $0.9$. Para ello en lugar de ir pidiendo al programa que calcule cada caso, lo vamos a hacer mediante una iteración. Le pediremos que realice unas operaciones un número determinado de veces.








```python
# primero definimos la formula general, e=0.9 y h=2 luego
altura= lambda n: 0.9**(2*n)*2  # ** denota elevado y * solo multiplicación
# usaremos for i in range(veces): para que repita esto, debes tabular lo que se repite.
#observa que comienza en 0
for i in range(10):
  print("la altura de la pelota en el rebote "+str(i)+" es ")
  print(altura(i))
```

Puedes observar que a medida que rebota, pierda altura, su coeficiente es menor que $1$ y por tanto cada vez el valor es menor.
¿Podrías saber cuál será su final, es decir, a que tiende esas alturas?
$$\lim_{n \to \infty}(0.9^{2n}\cdot 2)=0$$
Por muy alto que la dejemos caer, su límite volverá a ser cero. Pero, ¿qué ocurrirá si su coeficiente es mayor que 1?.


*   Una sucesión es **convergente**, si a medida que avanzamos en la sucesión, esta se estabiliza, decimos que su límite es el valor al que tiende.
*   Por el contrario, si la sucesión no se estabiliza, o tiende al infinito, tanto positivo como negativo, decimos que es **divergente**.

Usa el código anterior para investigar.  ¿Será divergente o convergente?. ¿De qué depende?.

```python
from decimal import *   # usamos esto para que nos muestre todos los decimales
altura= lambda n: 0.9**(2*n)*2  # ** denota elevado y * solo multiplicación
for i in range(10):
  print("%.20f" % Decimal(altura(i)))  # "%20f" % es para mostrarlo sin notación científica


print("\n Ahora con un coeficiente mayor que 1\n")
altura2= lambda n: 1.9**(2*n)*2  # ** denota elevado y * solo multiplicación
for i in range(10):
  print("%.10f" % Decimal(altura2(i)))
```

# Sucesiones recurrentes.

Son aquellas en las que sus términos se obtienes a partir de los anteriores aplicando una determinada regla.

Por ejemplo, dados los números, 3 y 4 obtenemos el siguientes sumando los dos anteriores y multiplicando por dos.

$$3,4,14,36,\dots$$

Vamos aprender ahora como podemos generar esta sucesión y obtener cualquier término. Observa que ahora no tenemos el término general, una fórmula que nos proporcione en base a la posición el valor que ocupa.  

Aprendemos a iterar en Python.


```python
#definimos los dos primeros términos
primero=3
segundo=4

#generamos la regla, para ello le pedimos al usario hasta donde quiere llegar, a que posición a partir de la tercera.
pos = int(input("Introduzca cuál es el valor de la posición cuyo valor desea conococer: "))

#imprimimos los dos primeros
print(str(primero),",",str(segundo),",",end="")
#es momento de iterar, de repetir un proceso muchas veces
# restamos 2 a la posición para que lo repita solo tres veces, pues los dos primeros ya los tiene
for i in range(pos-2):
 tercero=2*(primero+segundo)
 print(str(tercero),",",end="")
 primero=segundo # el primero pasa a ser el segundo, y el segundo será ahora el tercero.
 segundo=tercero
```
